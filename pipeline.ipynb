# Solar filament instance segmentation
Run YOLO11s-seg, refine each proposal with the two-channel U-Net, then apply the instance cleanup and export disjoint COCO RLE masks. The included checkpoints reproduce the selected experiment.

```python
from pathlib import Path
from reproduce import load, run_image
from predict_submission import export_csv

IMAGES = Path("data/test/images")
OUTPUT = Path("submission.csv")
VARIANT = "boundary-vote"
import cv2
cv2.setNumThreads(1)
recipe, detector, refiner = load()
```

```python
paths = sorted(path for path in IMAGES.iterdir() if path.suffix.lower() in {".jpg", ".jpeg", ".png"})
assert paths
predictions = {}
for path in paths:
    masks = run_image(path, recipe, detector, refiner, VARIANT)
    predictions[path.name] = {"file_name": path.name, "masks": masks}
```

```python
export_csv(OUTPUT, predictions)
```
